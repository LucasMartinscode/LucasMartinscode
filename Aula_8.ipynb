# Salvando no postgrees

```python
import pandas as pd
from sqlalchemy import create_engine

# Configurações de conexão com o PostgreSQL
db_user = 'postgres'
db_password = 'postgrespassword'
db_host = 'localhost'
db_port = '5431'
db_name = 'mydatabase'

# Criar a engine de conexão
engine = create_engine(f'postgresql+psycopg2://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}')

# Lista de arquivos CSV a serem importados
csv_files = [
    'C:\\Users\\LucasMartins\\OneDrive - SUPERCAMPO S.A\\Desktop\\Eng de dados\\awari-engenharia-de-dados-docker-main\\awari-engenharia-de-dados-docker-main\\exercicios\\municipios-estados\\csv\\municipios.csv',
    'C:\\Users\\LucasMartins\\OneDrive - SUPERCAMPO S.A\\Desktop\\Eng de dados\\awari-engenharia-de-dados-docker-main\\awari-engenharia-de-dados-docker-main\\exercicios\\municipios-estados\\csv\\estados.csv'
]

# Nomes das tabelas correspondentes no PostgreSQL
table_names = ['municipios', 'estados']

# Importar cada arquivo CSV para o PostgreSQL
for csv_file, table_name in zip(csv_files, table_names):
    # Carregar o CSV em um DataFrame
    df = pd.read_csv(csv_file)

    # Inserir os dados no PostgreSQL
    df.to_sql(table_name, engine, if_exists='replace', index=False)
    print(f"Dados do arquivo {csv_file} importados com sucesso para a tabela {table_name}!")
```
